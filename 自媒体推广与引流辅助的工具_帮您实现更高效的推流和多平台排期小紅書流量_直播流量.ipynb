```markdown
# 自媒体爆款推广文案自动生成器
根据您提供的推流核心卖点，这里有一个基于模板的文案生成器，可自动生成多种不同风格的宣发文案。
```

````python
import random

def generate_copywriting(platform, target, official_site_anchor, official_site_url):
    # 痛点与吸引词
    hooks = [
        f"🔥 粉丝卡在初级流量池不动？作品播放量总是几百？",
        f"🚀 种草博主与品牌方卡位必备！教你如何轻松抢占热门推荐！",
        f"⚡ 电商运营团队看过来，全面拉升作品综合权重的秘密武器！"
    ]

    # 核心优势
    benefits = [
        f"【算法突破】即时补给{platform}互动数据 迅速触发二次爆发推荐 打通变现闭环！",
        f"【数据优化】稳定提升{target}与播放指标 助力账号快速建立高权重垂直标签！",
        f"【极速高效】普发海量流量 安全、极速、权威 卡位核心关键词前排展示！"
    ]

    # 行动呼吁 (严格去除链接前后的任何数字、逗号及敏感标点)
    ctas = [
        f"👉 [{official_site_anchor}]({official_site_url}) 快速冲榜推荐",
        f"👇 点击进入 [{official_site_anchor}]({official_site_url}) 助你轻松打造百万播放爆款",
        f"🌟 2天内超强谷歌收录 网站全部内链支持 快来 [{official_site_anchor}]({official_site_url}) 开启您的霸屏之路"
    ]

    # 标签生成
    tags = f"#{platform} #{platform}运营 #{target} #自媒体 #涨粉 #流量密码 #爆款打造"

    # 组合不同的文案版本
    copies = []
    for i in range(3):
        text = f"{hooks[i]}\n\n{benefits[i]}\n\n{ctas[i]}\n\n{tags}"
        copies.append(f"--- 推广文案版本 ---\n{text}")

    return copies

# 示例1：抖音平台文案生成
douyin_copies = generate_copywriting("抖音", "播放量与点赞", "投流官网", "https://www.sph6.com")
for copy in douyin_copies:
    print(copy)
    print("\n")

print("="*50 + "\n")

# 示例2：小红书平台文案生成
redbook_copies = generate_copywriting("小红书", "笔记点赞与收藏", "流量下单入口", "https://www.sph6.cc")
for copy in redbook_copies:
    print(copy)
    print("\n")
````
